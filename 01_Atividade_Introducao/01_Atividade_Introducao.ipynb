**1 - O que é uma imagem digital?**

Uma imagem digital é uma função bidimensional discretizada $f(x,y)$, onde $x$ e $y$ denotam as coordenadas espaciais e o valor de $f$ em qualquer ponto $(x,y)$ é proporcional ao brilho da imagem naquele ponto. Ela é representada por uma matriz bidimensional com valores de intensidades, onde cada elemento dessa matriz é chamado de pixel ("picture element").

**2 - O que é visão computacional e por que é importante?**

A visão computacional é um subcampo da Inteligência Artificial que treina computadores para processar, analisar e interpretar dados visuais (imagens e vídeos), imitando a visão humana. Ela se ocupa da extração, análise e compreensão automáticas de informações úteis a partir de imagens.

É importante porque permite que máquinas realizem tarefas de forma autônoma (como dirigir ou inspecionar qualidade), além de superar as limitações do sistema visual humano, que é mais qualitativo do que quantitativo. Sistemas de visão computacional podem fazer medições precisas no mundo físico e extrair informações que humanos não conseguem perceber.

**3 - Quais são os passos fundamentais em processamento de imagens? Explique cada um deles.**

Os passos fundamentais em processamento de imagens são:

1. **Aquisição de imagens:** É o processo de obter a imagem digital usando sensores de imageamento, como câmeras, drones, tomógrafos, etc.
2. **Pré-processamento:** Consiste em melhorar a imagem para aumentar as chances de sucesso das etapas seguintes. Exemplos incluem realce de contraste e remoção de ruídos.
3. **Segmentação:** Divide uma imagem de entrada em suas partes ou objetos constituintes, sendo uma das tarefas mais difíceis do processamento.
4. **Representação e descrição:** A representação transforma os dados da imagem em um formato adequado para processamento computacional (ex: codificação da borda). A descrição extrai características que resultam em informações quantitativas de interesse (ex: área, perímetro).
5. **Reconhecimento e interpretação:** O reconhecimento atribui um rótulo a um objeto, enquanto a interpretação atribui um significado a um conjunto de objetos, envolvendo uma análise semântica mais complexa.

**4 - Explique os processos de amostragem e quantização para obtenção de imagens digitais. Explique também de que forma esses processos afetam a qualidade de uma imagem digital.**

Para que uma função contínua seja processada computacionalmente, ela precisa ser digitalizada no espaço e na amplitude:

- **Amostragem:** É a digitalização das coordenadas espaciais $(x,y)$. Os valores das coordenadas passam a ser discretos (inteiros e positivos). Quanto maior a amostragem, mais detalhes a imagem terá, mas exigirá maior espaço de armazenamento. Uma amostragem baixa causa perda de detalhes e efeito de pixelização.
- **Quantização:** É a digitalização da amplitude, ou seja, dos níveis de intensidade (ex: níveis de cinza). Os valores passam a ser discretos. A quantização define o número de tons que cada pixel pode assumir. Menos níveis de quantização reduzem o tamanho do arquivo, mas causam perda de informação e qualidade visual (menos tons de cinza ou cores disponíveis).

**5 - Abra uma imagem em escala de cinza e faça as quantizações para: 128, 64, 32, 16, 8 e 2 níveis de cinza. Mostre na tela tanto a imagem de entrada como todas as quantizações.**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Substitua 'gato.jpg' pelo caminho da sua imagem
img = cv2.imread('gato.jpg', cv2.IMREAD_GRAYSCALE)

if img is not None:
    niveis = [128, 64, 32, 16, 8, 2]
    
    plt.figure(figsize=(15, 10))
    
    plt.subplot(3, 3, 1)
    plt.imshow(img, cmap='gray')
    plt.title('Original (256 níveis)')
    plt.axis('off')
    
    for i, nivel in enumerate(niveis):
        fator = 256 / nivel
        img_quantizada = np.uint8(np.floor(img / fator) * fator)
        
        plt.subplot(3, 3, i + 2)
        plt.imshow(img_quantizada, cmap='gray', vmin=0, vmax=255)
        plt.title(f'Quantizada: {nivel} níveis')
        plt.axis('off')
        
    plt.tight_layout()
    plt.show()
else:
    print("Imagem não encontrada. Defina o caminho correto.")
```

**6 - Leia uma imagem colorida em RGB, crie variáveis para cada canal do RGB e mostre na tela cada um separadamente. Logo em seguida percorra e ache a maior e menor intensidade dos pixels de cada canal.**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img_colorida = cv2.imread('gato.jpg')

if img_colorida is not None:
    img_rgb = cv2.cvtColor(img_colorida, cv2.COLOR_BGR2RGB)
    
    r, g, b = cv2.split(img_rgb)
    
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 4, 1)
    plt.imshow(img_rgb)
    plt.title('Imagem Original RGB')
    plt.axis('off')
    
    plt.subplot(1, 4, 2)
    plt.imshow(r, cmap='Reds')
    plt.title('Canal Red (Vermelho)')
    plt.axis('off')
    
    plt.subplot(1, 4, 3)
    plt.imshow(g, cmap='Greens')
    plt.title('Canal Green (Verde)')
    plt.axis('off')
    
    plt.subplot(1, 4, 4)
    plt.imshow(b, cmap='Blues')
    plt.title('Canal Blue (Azul)')
    plt.axis('off')
    
    plt.show()
    
    print(f"Canal R - Maior: {np.max(r)}, Menor: {np.min(r)}")
    print(f"Canal G - Maior: {np.max(g)}, Menor: {np.min(g)}")
    print(f"Canal B - Maior: {np.max(b)}, Menor: {np.min(b)}")
else:
    print("Imagem não encontrada.")
```

7 - Leia uma imagem em escala de cinza e atribua o valor zero a todos os pixels cujo vizinho à direita tenha o mesmo valor de intensidade. Mostre a imagem de entrada e a imagem resultante do processo na tela.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('gato.jpg', cv2.IMREAD_GRAYSCALE)

if img is not None:
    img_resultante = img.copy()
    linhas, colunas = img.shape
    
    for i in range(linhas):
        for j in range(colunas - 1):
            if img[i, j] == img[i, j + 1]:
                img_resultante[i, j] = 0
                
    plt.figure(figsize=(10, 5))
    
    plt.subplot(1, 2, 1)
    plt.imshow(img, cmap='gray')
    plt.title('Imagem Original')
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.imshow(img_resultante, cmap='gray')
    plt.title('Imagem Resultante')
    plt.axis('off')
    
    plt.show()
else:
    print("Imagem não encontrada.")
```

8 - Leia uma imagem em escala de cinza e aplique dois métodos de binarização:

*   Limiarização com valor fixo
*   Limiarização automática utilizando o método de Otsu

Mostre na tela:

*   imagem original
*   resultado com limiar fixo
*   resultado com Otsu

Compare visualmente os resultados e informe o valor de limiar encontrado pelo método de Otsu.

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('gato.jpg', cv2.IMREAD_GRAYSCALE)

if img is not None:
    valor_limiar = 127
    _, img_limiar_fixo = cv2.threshold(img, valor_limiar, 255, cv2.THRESH_BINARY)
    
    limiar_otsu, img_otsu = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.imshow(img, cmap='gray')
    plt.title('Imagem Original')
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    plt.imshow(img_limiar_fixo, cmap='gray')
    plt.title(f'Limiar Fixo (v={valor_limiar})')
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    plt.imshow(img_otsu, cmap='gray')
    plt.title(f'Método de Otsu (Limiar={limiar_otsu})')
    plt.axis('off')
    
    plt.show()
    
    print(f"O valor de limiar encontrado automaticamente pelo método de Otsu foi: {limiar_otsu}")
else:
    print("Imagem não encontrada.")
```

9 - Leia uma imagem colorida e converta-a para escala de cinza de três formas:

*   Utilizando a função pronta do OpenCV
*   Utilizando a fórmula matemática de conversão manual
*   Utilizando a fórmula matemática de conversão manual sem pesos nos canais

FÓRMULA: CINZA=0.2125R+0.7154G+0.0721B

Mostre na tela:

*   imagem colorida original
*   imagem em escala de cinza obtida pelo OpenCV
*   imagem em escala de cinza obtida manualmente
*   imagem em escala de cinza obtida manualmente sem pesos nos canais

Compare visualmente os resultados.
Por fim, pesquise e informe quais valores de pesos são usados na função pronta do Opencv.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img_colorida = cv2.imread('gato.jpg')

if img_colorida is not None:
    img_rgb = cv2.cvtColor(img_colorida, cv2.COLOR_BGR2RGB)
    
    img_cv2_gray = cv2.cvtColor(img_colorida, cv2.COLOR_BGR2GRAY)
    
    R = img_rgb[:,:,0].astype(np.float32)
    G = img_rgb[:,:,1].astype(np.float32)
    B = img_rgb[:,:,2].astype(np.float32)
    
    img_manual = 0.2125 * R + 0.7154 * G + 0.0721 * B
    img_manual = np.uint8(img_manual)
    
    img_manual_sem_pesos = (R + G + B) / 3
    img_manual_sem_pesos = np.uint8(img_manual_sem_pesos)
    
    plt.figure(figsize=(20, 5))
    
    plt.subplot(1, 4, 1)
    plt.imshow(img_rgb)
    plt.title('Original Colorida')
    plt.axis('off')
    
    plt.subplot(1, 4, 2)
    plt.imshow(img_cv2_gray, cmap='gray')
    plt.title('Cinza - OpenCV')
    plt.axis('off')
    
    plt.subplot(1, 4, 3)
    plt.imshow(img_manual, cmap='gray')
    plt.title('Cinza - Manual c/ Pesos')
    plt.axis('off')
    
    plt.subplot(1, 4, 4)
    plt.imshow(img_manual_sem_pesos, cmap='gray')
    plt.title('Cinza - Manual s/ Pesos (Média)')
    plt.axis('off')
    
    plt.show()
    
    print("Comparação Visual:")
    print("A imagem calculada sem pesos fica diferente porque o olho humano não é igualmente sensível às cores primárias.")
    print("\nPesos usados na função pronta do OpenCV (padrão Rec. 601):")
    print("Y = 0.299*R + 0.587*G + 0.114*B")
else:
    print("Imagem não encontrada.")
```
