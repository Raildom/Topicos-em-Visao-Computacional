## 1) O que são filtros espaciais de suavização? Quais são suas principais aplicações?

Filtros espaciais de suavização (passa-baixa) são técnicas no domínio espacial que operam diretamente sobre a informação dos pixels por meio do deslizamento de uma máscara (kernel) e operações de convolução, calculando a soma ponderada dos pixels na vizinhança para substituir o valor do pixel central.

**Principais aplicações:**
* **Borrar a imagem:** suavizar características indesejadas, como detalhes finos ou variações abruptas de intensidade.
* **Reduzir o ruído:** eliminar perturbações indesejadas na imagem (como ruído gaussiano ou ruído sal-e-pimenta).
* **Pré-processamento:** remover artefatos e preparar a imagem para análises subsequentes (como detecção de bordas ou reconhecimento de objetos).

```python
# Questão 3 — Importação das bibliotecas e carregamento da imagem (gato.jpg) em nível de cinza
import numpy as np
import matplotlib.pyplot as plt
import math
import os
from PIL import Image

# Carregamento da imagem própria em escala de cinza (1 canal)
caminho_imagem = 'gato.jpg'
if not os.path.exists(caminho_imagem):
    caminho_imagem = '04_Atividade_Técnicas_de_Filtragem_Suavização/gato.jpg'

img_pil = Image.open(caminho_imagem).convert('L')
img_real = np.array(img_pil, dtype=np.float64)

print(f'Imagem carregada: {caminho_imagem} (Dimensões: {img_real.shape})')

plt.figure(figsize=(4, 4))
plt.imshow(img_real, cmap='gray', vmin=0, vmax=255)
plt.title('Imagem de Entrada (gato.jpg)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Questão 3 — Implementação manual dos filtros espaciais de suavização (sem bibliotecas prontas)
def pad_replicate(img, pad):
    return np.pad(img, pad_width=pad, mode='edge')

def filtro_media(img, mask_size=3):
    rows, cols = img.shape
    pad = mask_size // 2
    output = np.zeros_like(img)
    img_pad = pad_replicate(img, pad)
    for i in range(rows):
        for j in range(cols):
            ii, jj = i + pad, j + pad
            vizinhanca = img_pad[ii - pad : ii + pad + 1, jj - pad : jj + pad + 1]
            output[i, j] = int(np.sum(vizinhanca) / (mask_size * mask_size))
    return output

def filtro_mediana(img, mask_size=3):
    rows, cols = img.shape
    pad = mask_size // 2
    output = np.zeros_like(img)
    img_pad = pad_replicate(img, pad)
    for i in range(rows):
        for j in range(cols):
            ii, jj = i + pad, j + pad
            vizinhanca = img_pad[ii - pad : ii + pad + 1, jj - pad : jj + pad + 1].flatten()
            vizinhanca_ord = sorted(vizinhanca)
            output[i, j] = int(vizinhanca_ord[len(vizinhanca_ord) // 2])
    return output

def gerar_kernel_gaussiano(tamanho, sigma):
    pad = tamanho // 2
    kernel = np.zeros((tamanho, tamanho), dtype=np.float64)
    for i in range(tamanho):
        for j in range(tamanho):
            x = i - pad
            y = j - pad
            kernel[i, j] = math.exp(-(x**2 + y**2) / (2 * sigma**2))
    kernel /= kernel.sum()
    return kernel

def filtro_gaussiano(img, mask_size=3, sigma=1.0):
    rows, cols = img.shape
    pad = mask_size // 2
    output = np.zeros_like(img)
    kernel = gerar_kernel_gaussiano(mask_size, sigma)
    img_pad = pad_replicate(img, pad)
    for i in range(rows):
        for j in range(cols):
            ii, jj = i + pad, j + pad
            vizinhanca = img_pad[ii - pad : ii + pad + 1, jj - pad : jj + pad + 1]
            output[i, j] = int(np.clip(np.sum(vizinhanca * kernel), 0, 255))
    return output
```

```python
# Questão 3a — Filtro da Média (máscaras 3x3, 5x5 e 7x7)
tamanhos = [3, 5, 7]
resultados_media = [filtro_media(img_real, m) for m in tamanhos]

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
axes[0].imshow(img_real, cmap='gray', vmin=0, vmax=255)
axes[0].set_title('Imagem de Entrada (Original)', fontsize=11)
axes[0].axis('off')
for ax, res, m in zip(axes[1:], resultados_media, tamanhos):
    ax.imshow(res, cmap='gray', vmin=0, vmax=255)
    ax.set_title(f'Filtro da Média {m}x{m}', fontsize=11)
    ax.axis('off')
plt.suptitle('Questão 3a — Filtro da Média (3x3, 5x5, 7x7)', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Questão 3b — Filtro da Mediana (máscaras 3x3, 5x5 e 7x7)
resultados_mediana = [filtro_mediana(img_real, m) for m in tamanhos]

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
axes[0].imshow(img_real, cmap='gray', vmin=0, vmax=255)
axes[0].set_title('Imagem de Entrada (Original)', fontsize=11)
axes[0].axis('off')
for ax, res, m in zip(axes[1:], resultados_mediana, tamanhos):
    ax.imshow(res, cmap='gray', vmin=0, vmax=255)
    ax.set_title(f'Filtro da Mediana {m}x{m}', fontsize=11)
    ax.axis('off')
plt.suptitle('Questão 3b — Filtro da Mediana (3x3, 5x5, 7x7)', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Questão 3c — Filtro Gaussiano com sigma=1 (máscaras 3x3, 5x5 e 7x7)
sigma = 1.0
resultados_gauss = [filtro_gaussiano(img_real, m, sigma) for m in tamanhos]

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
axes[0].imshow(img_real, cmap='gray', vmin=0, vmax=255)
axes[0].set_title('Imagem de Entrada (Original)', fontsize=11)
axes[0].axis('off')
for ax, res, m in zip(axes[1:], resultados_gauss, tamanhos):
    ax.imshow(res, cmap='gray', vmin=0, vmax=255)
    ax.set_title(f'Gaussiano {m}x{m} (sigma={sigma})', fontsize=11)
    ax.axis('off')
plt.suptitle(f'Questão 3c — Filtro Gaussiano (sigma={sigma})', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Questão 3c — Filtro Gaussiano variando o valor de sigma (0.5, 1.0, 2.0, 3.0)
sigmas = [0.5, 1.0, 2.0, 3.0]
mask_size_fixo = 7
resultados_sigmas = [filtro_gaussiano(img_real, mask_size_fixo, s) for s in sigmas]

fig, axes = plt.subplots(1, 5, figsize=(20, 4))
axes[0].imshow(img_real, cmap='gray', vmin=0, vmax=255)
axes[0].set_title('Original', fontsize=11)
axes[0].axis('off')
for ax, res, s in zip(axes[1:], resultados_sigmas, sigmas):
    ax.imshow(res, cmap='gray', vmin=0, vmax=255)
    ax.set_title(f'Gaussiano {mask_size_fixo}x{mask_size_fixo}\nsigma={s}', fontsize=11)
    ax.axis('off')
plt.suptitle(f'Questão 3c — Variação de sigma (máscara {mask_size_fixo}x{mask_size_fixo})', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```
